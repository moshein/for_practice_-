```python
def search(root, data):
        # if walay root, return False
        if root is None:
            return False
        # if nakita ang data
        if root.data == data:
            return True
        # icheck tanan anak sa node
        for child in root.children:
            # recursive search sa children
            if search(child, data):
                return True
        # if way nakita
        return False

def delete(root, node):
  if not root:
    return

  for child in root.children: ## mag loop sya hekhek
    if child.data == node:
      root.children.extend(child.children) ## iconnect ata niya para dili ma-remove yung children/grandchildren once iremove yung node (para ma retain parin yung children sa node na iremove)
      root.children.remove(child) ## nag remove na
      return True

    if delete(child, node):
      return True

    return False

def delete_subtree(root, node):
  # aku
  if not root:
    return False

  for child in root.children: ## nag loop sya againn
    if child.data == node:
      root.children.remove(child) ## if ma-see niya ang node na iremove, iremove na dayun niya (including children (of the node to be removed))
      return True

    if delete_subtree(child, node): ## iask lang niya yung child subtree na itry delete yung node (actually dili pa nako fully comprehend huhuhuh ako istudy later T^T)
      return True

  return False


def find_parent(root, node_data):
        # Pangitaa ang parent sa children nga node
        # Start sa root if wala pay current node.
        if current is None:
            current = root

        # If empty ang tree, then obviously, wala siyay parent.
        if current is None:
            return None

        # I-scan ang tanang children sa current node.
        for child in current.children:
            # If ang data sa child kay match sa node_data, then 'current' is the parent!
            if child.data == node_data:
                return current # Found it! Mao ni ang parent.

            # If dili, subay pa ta down the tree recursively.
            res = find_parent(node_data, child)
            if res:
                return res

        # If nahuman na ang loop unya wala gyud, return None.
        return None

```
