```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return
        
        temp = self.head 

        while temp.next:
            temp = temp.next
        
        temp.next = new_node

    def pop(self):
        temp = self.head
        
        while temp.next.next:
            temp = temp.next
        
        popped_value = temp.next
        temp.next = None
        return popped_value.data
    
    def remove(self, key):
        temp = self.head

        if self.head.data == key:
            self.head = temp.next
        
        while temp.next:
            if temp.next.data == key:
                temp.next = temp.next.next

    def sort(self):
        temp1 = self.head

        while temp1:
            temp2 = temp1.next

            while temp2:
                if temp1.data > temp2.data:
                    temp1.data, temp2.data = temp2.data, temp1.data
                temp2 = temp2.next
            temp1 = temp1.next

    def search(self, key):
        temp = self.head
        position = 0
        
        while temp:
            if temp.data == key:
                print(f"found [{key}] at location [{position}] mimiao miao")
            position += 1
            temp = temp.next

    def display(self):
        temp = self.head 

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next

        print("nonee")

```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None
        self.count = 0 ### dapat daw dili same ang iya name sa function
        # self.max = max
    
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        self.count += 1
    
    def peek(self):
        return self.top.data
    
    def pop(self):
        popped_value = self.top.data

        temp = self.top
        self.top = temp.next

        self.count -= 1
        
        return popped_value 
    
    # def count(self):
    #     return self.count.data 
            ### dapat daw dili same ang iya name sa function
    
    def size(self):
        return self.count

    def display(self):
        temp = self.top

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        print("noneee")


```

```python
s = Stack()

s.push(1)
s.push(2)
s.push(3)
s.push(4)
s.display()

s.pop()
s.display()

s.size()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
    
    def enqueue(self, data):
        new_node = Node(data)

        if not self.front:
            self.front = self.rear = new_node
            return
        
        self.rear.next = new_node
        self.rear = new_node
    
    def dequeue(self):
        temp = self.front
        
        # v_to_be_removed = temp
        v_to_be_removed = temp.data ## kalimot ta mag put ug data hehe
        self.front = temp.next

        ## rawrawr kalimot ta mag put ug empty queue handling thingy

        if self.front is None:
            self.rear = None
            return

        return v_to_be_removed
    
    def display(self):
        temp = self.front

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        print("nonee")

        

```

```python
q = Queue()

q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.display()

q.dequeue()
q.display()
```
