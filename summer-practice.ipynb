```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return
        
        # temp = new_node

        temp = self.head

        while temp.next:
            temp = temp.next
        
        temp.next = new_node
    
    def pop(self):
        temp = self.head
        
        while temp.next.next:
            temp = temp.next
        
        popped_value = temp.next

        temp.next = None

        return popped_value
    
    def remove(self, key):
        temp = self.head

        if self.head.data == key:
            self.head = temp.next
            return
        
        while temp.next:
            if temp.next.data == key:
                temp.next = temp.next.next
                return
            temp = temp.next
        
    def sort_bubble(self):
        temp1 = self.head

        while temp1:
            temp2 = temp1.next

            while temp2:
                if temp1.data > temp2.data:
                    temp1.data, temp2.data = temp2.data, temp1.data
                temp2 = temp2.next
            temp1 = temp1.next

    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        print("none")

ll = LinkedList()

ll.append(1)
ll.append(2)
ll.append(4)
ll.append(3)
ll.append(2)
ll.append(7)
ll.display()

ll.sort_bubble()
ll.display()
```

## Rearranging an Array

Given an array of elements of length n, ranging from 0 to n - 1. All elements may not be present in the array. If the element is not present then there will be -1 present in the array. Rearrange the array such that arr[i] = i and if i is not present, store -1 at that place.

```python
'''
Input: arr[] = [-1, -1, 6, 1, 9, 3, 2, -1, 4, -1]
Output: [-1, 1, 2, 3, 4, -1, 6, -1, -1, 9]
Explanation: In range 0 to 9, all except 0, 5, 7 and 8 are present. Hence, we print -1 instead of them.

Input: arr[] = [0, 1, 2, 3, 4, 5] 
Output: [0, 1, 2, 3, 4, 5]
Explanation: In range 0 to 5, all number are present.
'''
```

```python
def modifyArray(ar, n):
    print("Initial array:", ar)

    for i in range(n):
        found = False
        for j in range(n):
            if ar[j] == i:
                print(f"Step: value {i} found at index {j}, swap with index {i}")
                ar[j], ar[i] = ar[i], ar[j]
                print(" Current array:", ar)
                found = True
                break
        if not found:
            print(f"Step: value {i} not found, leave index {i} for -1")

    print("After rearranging positions:", ar)

    for i in range(n):
        if ar[i] != i:
            print(f"Step: index {i} has {ar[i]}, set to -1")
            ar[i] = -1
            print(" Current array:", ar)

    print("Final array:", ar)

ar = [-1, -1, 6, 1, 9, 3, 2, -1, 4, -1]
n = len(ar)

modifyArray(ar, n)
```
