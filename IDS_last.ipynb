```python
import requests
from IPython.display import display, Markdown

try:
    url = 'https://reddit.com'

    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    with requests.Session() as session:
        connection = session.get(url, headers=headers)

        if connection.ok:
            display(
                Markdown(
                    f'<b>Success</b>: '
                    f'<u><i>{connection.status_code} {connection.reason}</i></u>'
                )
            )
        else:
            display(
                Markdown(
                    f'<b>Unexpected status code</b>: '
                    f'{connection.status_code} {connection.reason}'
                )
            )

except requests.exceptions.ConnectionError as e:
    display(Markdown(f'<b>Connection failed</b>: {e}'))
```
