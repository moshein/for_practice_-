```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return self.head
        
        temp = self.head

        while temp.next:
            temp = temp.next
        
        temp.next = new_node
    
    def pop(self):
        temp = self.head
        
        while temp.next.next:
            temp = temp.next
        
        temp.next = None
    
    def remove(self, key):
        temp = self.head

        if self.head.data == key:
            popped_val = temp.next
            self.head = temp.next
            return popped_val

        while temp.next:
            if temp.next.data == key:
                popped_val = temp.next
                temp.next = None
                return
            temp = temp.next
            
    
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        
        print("nonee")

ll = LinkedList()
ll.append(1)
ll.append(3)
ll.append(3)
ll.append(3)
ll.append(4)
ll.append(5)
ll.display()

ll.remove(3)
ll.display()

### uhm... unsa kahay mali ani ba AHAHHAHAHAHAHHAHAHA
        

```
