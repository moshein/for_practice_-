```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return
        
        temp = self.head

        while temp:
            temp = temp.next

        temp.next = new_node
    
    
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
    
    def enqueue(self, data):
        new_node = Node(data)

        if self.rear is None:
            self.front = self.rear = new_node
            return
        
        # self.rear.next = new_node
        # self.rear = new_node 

        new_node.next = self.rear
        self.rear = new_node

    def display(self):
        temp = self.front

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        print("None")
```

```python
q = Queue()

q.enqueue(1)
q.enqueue(2)
q.enqueue(4)

q.display()
```
