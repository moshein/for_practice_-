```python
def large_num(num_lis):
    '''
    Find the largest number without using max()
    '''

    largest = 0
    
    for i in num_lis:
        if i > largest:
            largest = i

    return largest


num_lis = [12, 45, 7, 89, 23, 56, 3, 91, 34]

large_num(num_lis)
```

```python
def student_stat(lis):
    '''
    Using the given list, calculate the:
    Mean
    Highest score
    Lowest score
    Number of students who passed
    Number of students who failed
    Passing percentage
    '''

    count_stud = 0
    sum_scores = 0
    highest_score = 0
    lowest_score = lis[0]
    num_passed = 0
    num_failed = 0

    # mean
    for i in lis:
        count_stud += 1
        sum_scores += i

    class_mean = sum_scores / count_stud

    # highest score
    for i in lis:
        if i > highest_score:
            highest_score = i 

    # lowest score
    for i in lis:
        if i < lowest_score:
            lowest_score = i

    # number of students who passed
    for i in lis:
        if i >= 75:
            num_passed += 1
        else:
            num_failed += 1 # number of students who failed

    # passing percentage
    pass_percent = (num_passed / count_stud)*100

    return print(f"Mean: {class_mean}\nHighest score: {highest_score}\nLowest score: {lowest_score}\nNumber of students who passed: {num_passed}\nNumber of students who failed: {num_failed}\nPassing percentage: {pass_percent}")

lis = [85, 92, 78, 90, 88, 76, 95, 89, 84, 91]

student_stat(lis)
```
