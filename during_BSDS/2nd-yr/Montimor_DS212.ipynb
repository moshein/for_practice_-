```python
class Assignment:
    def __init__(self, course_name, assgt_details):
        self.course_name = course_name
        self.assgt_details = assgt_details

class Course:
    def __init__(self, course_name):
        self.course_name = course_name
        self.assgt = []
    
    def add_assgt(self, assgt: Assignment):
        # self.assgt.append(Assignment) THIS IS WRONG WRONG PAG CALL RAWRWARR
        self.assgt.append(assgt)

class Person:
    def __init__(self, schoolID, name):
        self.schoolID = schoolID
        self.name = name

class Teacher(Person):
    def __init__(self, schoolID, name):
        super().__init__(schoolID, name)
    
    def discuss_topic(self, a: Course):
        # return f"teacher is discussing {Course.course_name}" WRONG DAW KAY CLASS BLUEPRINT PAMAN DAW SYA
        return f"teacher is discussing {a.course_name}"

class Student(Person):
    def __init__(self, schoolID, name):
            super().__init__(schoolID, name)
        
    def attend_class(self, a: Course):
        return f"student is attending class: {a.course_name}"
    
    def make_assgt(self, assgt: Assignment):
        return f"student is making assignment: {assgt.assgt_details}"

    


    
        
```

```python
a1 = Assignment("SUB1", "assgt1")
a2 = Assignment("SUB2", "assgt2")

course1 = Course("SUB1")

course1.add_assgt(a1)
course1.add_assgt(a2)

print(course1.course_name)

for i in course1.assgt:
    print(i.course_name, i.assgt_details)
```

```python
t = Teacher("1212d", "Teacher A")

print(t.discuss_topic(course1)) 

s = Student("S999a", "Student B")
print(s.attend_class(course1))
print(s.make_assgt(a1))
```
