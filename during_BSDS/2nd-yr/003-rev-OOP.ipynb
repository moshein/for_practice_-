## Aggregation + Inheritance + Polymorphism

```python
class Player:
    def __init__(self, name: str):
        self._name = name

    def play(self):
        return f"{self._name} is playing"

class Guard(Player):
    def __init__(self, name: str):
        super().__init__(name)

    def play(self):
        return f"{self._name} is playing as the guard"

class Center(Player):
    def __init__(self, name):
        super().__init__(name)

    def play(self):
        return f"{self._name} is playing as the center"

class Team:
    def __init__(self, teamName: str):
        self._teamName = teamName
        self._players = []

    def add_player(self, player):
        self._players.append(player)
        return f"{player._name} is added to the team"

team = Team("Ateneo Eagles")

guard = Guard("alex")
center = Center("sam")

print(team.add_player(guard))
print(team.add_player(center))
```

```python
for player in team._players:
    print(player.play())
```

## Composition + Inheritance + Polymorphism

```python
class Room:
    def __init__(self, name):
        self._name = name

    def describe(self):
        return f"{self._name} is a room"

class Bedroom(Room):
    def __init__(self):
        super().__init__("Bedroom")

    def describe(self):
        return super().describe()

class Kitchen(Room):
    def __init__(self):
        super().__init__("Kitchen")

    def describe(self):
        return super().describe()

class Bathroom(Room):
    def __init__(self):
        super().__init__("Bathroom")

    def describe(self):
        return super().describe()

class House:
    def __init__(self, address):
        self._address = address
        self._rooms = []

    def add_room(self, roomi):
        self._rooms.append(roomi)
        return f"{roomi._name} is added to house"


house = House("123 Main Street")

print(house.add_room(Bedroom()))
print(house.add_room(Kitchen()))
print(house.add_room(Bathroom()))
```

```python
for i in range(len(house._rooms)):
    print(house._rooms[i]._name)
```

## Dependency + Inheritance + Polymorphism

```python
class Printer:
    def __init__(self, name: str):
        self._name = name

    def print(self):
        return f"{self._name} is printing"

class InkPrinter(Printer):
    def __init__(self):
        super().__init__("Ink Printer")

    def print(self):
        return super().print()

class LaserPrinter(Printer):
    def __init__(self):
        super().__init__("Laser Printer")

    def print(self):
        return super().print()

class Student:
    def __init__(self, name: str):
        self._name = name

    def print_file(self, printer, fileName: str):
        return f"{printer._name} is now printing {fileName}"

s = Student("miku")

# s.print_file(LaserPrinter, "report.pdf") ### mag error sya cuz wala paman naka-make ug instance sa class LaserPrinter

laser_printer = LaserPrinter()

s.print_file(laser_printer, "report.pdf")
```
