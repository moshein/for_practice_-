# Linked List using Linked List itself

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)

        if not self.head:
            self.head = new_node
            return

        temp = self.head

        while temp.next:
            temp = temp.next
        
        temp.next = new_node
    
    def pop(self):
        temp = self.head

        while temp.next.next:
            temp = temp.next

        popped_value = temp.next

        temp.next = None

        # a > b > c > d > e
        # mag stop sya kay d
        # why?
        # cuz temp.next.next is non existent man
        # thus temp.next == e
    
    def remove(self, key):
        temp = self.head

        if self.head.data == key:
            self.head = temp.next
            return

        while temp.next:
            if temp.next.data == key:
                temp.next = temp.next.next
                return
            temp = temp.next
    
    def sort(self):
        temp1 = self.head

        while temp1:
            temp2 = temp1.next

            while temp2:
                if temp1.data > temp2.data:
                    temp1.data, temp2.data = temp2.data, temp1.data
                temp2 = temp2.next
            temp1 = temp1.next
    
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end="  ->  ")
            temp = temp.next
        
        print("nonekeke")

        
```
